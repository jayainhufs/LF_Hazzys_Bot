# 05. 임베딩 모델 비교 (Gemini vs Local)

동일 query/document 에 대한 cosine similarity 를 빠르게 비교한다.

.env 의 EMBEDDING_PROVIDER / LOCAL_EMBEDDING_MODEL 을 바꿔가며 검사할 수 있다.


```python
import sys, pathlib
ROOT = pathlib.Path('..').resolve()
sys.path.insert(0, str(ROOT))

from src.evaluation.embedding_eval import cosine_similarity
from src.rag.embedder import Embedder

query = 'Meta 캠페인 세팅 시 ROAS 기준이 어떻게 되나요?'
docs = [
    'Meta 광고 캠페인 세팅 절차와 ROAS 기준 점검 가이드',
    '카카오 비즈보드 소재 검수 안내',
    '구글 키워드 매칭 옵션 정리',
]

for provider in ['gemini', 'local']:
    try:
        emb = Embedder(provider=provider)
        q = emb.embed_query(query)
        ds = emb.embed_documents(docs)
        print(f'\n## provider={provider} model={emb.model_name}')
        for d, v in zip(docs, ds):
            print(f'  sim={cosine_similarity(q, v):.4f}  ::  {d}')
    except Exception as e:
        print(f'provider={provider} 사용 불가: {e}')
```
