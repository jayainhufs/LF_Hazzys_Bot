# 04. 검색 품질 빠른 확인

특정 질문에 대해 retrieve 결과 / final_score 분포 / 카테고리 비율 등을 확인.


```python
import sys, pathlib
ROOT = pathlib.Path('..').resolve()
sys.path.insert(0, str(ROOT))

from src.rag.retriever import Retriever

retriever = Retriever()
results = retriever.retrieve('Meta 캠페인 세팅 시 ROAS 기준', top_k=8)
for r in results:
    print(f"{r.final_score:.4f}  {r.uploaded_category}/{r.content_type}  {r.file_name} :: {r.section_title}")
```
