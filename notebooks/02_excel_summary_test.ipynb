Gemini API 호출이 발생하므로 비용 주의.


```python
import sys, pathlib
ROOT = pathlib.Path('..').resolve()
sys.path.insert(0, str(ROOT))

from src.config import settings
from src.ingestion.excel_parser import parse_excel
from src.summarization.excel_summarizer import ExcelSummarizer
from src.utils.hash_utils import file_hash, short_hash
from src.utils.path_utils import iter_files

excel_dir = settings.raw_data_dir / settings.category_dirs['excel']
files = list(iter_files(excel_dir, exts={'.xlsx', '.xlsm'}))
print(files)
```

```python
if files:
    p = files[0]
    fhash = file_hash(p)
    document_id = f'doc_{short_hash(fhash, length=10)}'
    sections = parse_excel(p, document_id)
    summarizer = ExcelSummarizer()
    s = summarizer.summarize_section(
        document_id=document_id,
        file_name=p.name,
        sheet_name=sections[0]['metadata']['sheet_name'],
        raw_table_text=sections[0]['content'],
        table_range=sections[0]['metadata'].get('table_range'),
        source_raw_path=str(p),
    )
    print(s.summary_text[:1500])
```
